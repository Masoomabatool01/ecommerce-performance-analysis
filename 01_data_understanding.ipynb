```python
pd.read_sql_query("""
SELECT name
FROM sqlite_master
WHERE type = 'table';
""", conn)
```

```python
pd.read_sql_query("""
Select count(*) as total_orders
from orders;
""", conn)
```

```python
pd.read_sql_query("""
SELECT
    strftime('%Y', order_purchase_timestamp) AS year,
    COUNT(*) AS total_orders
FROM orders
GROUP BY year
ORDER BY year;
""", conn)
```

```python
pd.read_sql_query("""
SELECT
    ROUND(SUM(price), 2) AS total_revenue
FROM order_items;
""", conn)
```

```python
pd.read_sql_query("""
SELECT
    ROUND(SUM(price), 2) AS total_revenue,
    COUNT(DISTINCT order_id) AS total_orders,
    ROUND(SUM(price) / COUNT(DISTINCT order_id), 2) AS average_order_value
FROM order_items;
""", conn)
```

```python
kpi = pd.read_sql_query("""
SELECT
    COUNT(DISTINCT order_id) AS total_orders,
    ROUND(SUM(price), 2) AS product_revenue,
    ROUND(SUM(price) / COUNT(DISTINCT order_id), 2) AS average_order_value,
    COUNT(DISTINCT product_id) AS unique_products,
    COUNT(DISTINCT seller_id) AS unique_sellers
FROM order_items;
""", conn)

kpi
```

```python
pd.read_sql_query("""
SELECT
    p.product_category_name AS category,
    COUNT(DISTINCT oi.order_id) AS orders,
    ROUND(SUM(oi.price), 2) AS revenue
FROM order_items oi
LEFT JOIN products p
    ON oi.product_id = p.product_id
WHERE p.product_category_name IS NOT NULL
GROUP BY p.product_category_name
ORDER BY revenue DESC
LIMIT 10;
""", conn)
```

```python
order_payment.to_sql("order_payment", conn, if_exists="replace", index=False)

print(pd.read_sql_query("""
SELECT name
FROM sqlite_master
WHERE type = 'table';
""", conn))
```

```python
# 1. Payment Methods
payment_analysis = pd.read_sql_query("""
SELECT
    payment_type,
    COUNT(*) AS payment_records,
    ROUND(SUM(payment_value), 2) AS payment_value
FROM order_payment
GROUP BY payment_type
ORDER BY payment_value DESC;
""", conn)

# 2. Order Status
status_analysis = pd.read_sql_query("""
SELECT
    order_status,
    COUNT(*) AS orders
FROM orders
GROUP BY order_status
ORDER BY orders DESC;
""", conn)

# 3. Delivery Performance
delivery_analysis = pd.read_sql_query("""
SELECT
    ROUND(AVG(
        julianday(order_delivered_customer_date)
        - julianday(order_purchase_timestamp)
    ), 2) AS avg_delivery_days,
    
    ROUND(AVG(
        julianday(order_estimated_delivery_date)
        - julianday(order_purchase_timestamp)
    ), 2) AS avg_estimated_days
FROM orders
WHERE order_delivered_customer_date IS NOT NULL;
""", conn)

payment_analysis, status_analysis, delivery_analysis
```

```python
pd.read_sql_query("""
SELECT 
    review_score,
    COUNT(*) AS reviews
FROM order_reviews
GROUP BY review_score
ORDER BY review_score;
""", conn)
```

```python
pd.read_sql_query("""
SELECT 
    ROUND(AVG(review_score), 2) AS avg_review_score
FROM order_reviews;
""", conn)
```

```python
pd.read_sql_query("""
SELECT
    CASE
        WHEN review_score >= 4 THEN 'Positive'
        WHEN review_score <= 2 THEN 'Negative'
        ELSE 'Neutral'
    END AS sentiment,
    COUNT(*) AS reviews
FROM order_reviews
GROUP BY sentiment
ORDER BY reviews DESC;
""", conn)
```
